# KS plots

```python
import numpy as np
import matplotlib.pyplot as plt

cmap_heat = plt.colormaps["RdBu"].reversed()
cmap_heatR = plt.colormaps["Reds"]
cmap_heatB = plt.colormaps["Blues"].reversed()

Lfact = 7 # domain length factor
Ldom = Lfact * np.pi # domain length
dt = 0.25

Ntrue = 563 # time snapshots
Nq = 64 # delays
L = 2048 # spectral resolution

N = Ntrue - Nq + 1 # temporal space dimension
M = 64 # spatial space dimension
NM = N * M # product space dimension

nvec = np.arange(N)
xvec = Ldom * np.arange(-int(M/2), int(M/2))/M
```

## Eigenfunction basis -- dilution

```python
Phi1 = np.load(f"../data/ks_phi1_{NM}_{L}_{Lfact}_50.npy")
```

```python
phix = np.asarray([1, 2, 3, 4, 5, 6])

f0 = Phi1[:, phix[0]].reshape((N, M), order="C")
f1 = Phi1[:, phix[1]].reshape((N, M), order="C")
f2 = Phi1[:, phix[2]].reshape((N, M), order="C")
f3 = Phi1[:, phix[3]].reshape((N, M), order="C")
f4 = Phi1[:, phix[4]].reshape((N, M), order="C")
f5 = Phi1[:, phix[5]].reshape((N, M), order="C")

fig, ((ax0, ax1, ax2), (ax3, ax4, ax5)) = plt.subplots(2, 3, layout="tight", figsize=[6.4, 6.4])

pcm0 = ax0.pcolormesh(xvec, nvec, f0, cmap=cmap_heat)
ax0.tick_params(axis="x", length=0)
ax0.tick_params(axis="y", length=0)
ax0.set_ylabel("Sample")
fig.colorbar(pcm0, ax=ax0, location="top", pad=0.01)

pcm1 = ax1.pcolormesh(xvec, nvec, f1, cmap=cmap_heat)
ax1.tick_params(axis="x", length=0)
ax1.tick_params(axis="y", length=0)
fig.colorbar(pcm1, ax=ax1, location="top", pad=0.01)

pcm2 = ax2.pcolormesh(xvec, nvec, f2, cmap=cmap_heat)
ax2.tick_params(axis="x", length=0)
ax2.tick_params(axis="y", length=0)
fig.colorbar(pcm1, ax=ax2, location="top", pad=0.01)

pcm3 = ax3.pcolormesh(xvec, nvec, f3, cmap=cmap_heat)
ax3.tick_params(axis="x", length=0)
ax3.tick_params(axis="y", length=0)
ax3.set_xlabel("Space")
ax3.set_ylabel("Sample")
fig.colorbar(pcm3, ax=ax3, location="top", pad=0.01)

pcm4 = ax4.pcolormesh(xvec, nvec, f4, cmap=cmap_heat)
ax4.tick_params(axis="x", length=0)
ax4.tick_params(axis="y", length=0)
fig.colorbar(pcm4, ax=ax4, location="top", pad=0.01)

pcm5 = ax5.pcolormesh(xvec, nvec, f5, cmap=cmap_heat)
ax5.tick_params(axis="x", length=0)
ax5.tick_params(axis="y", length=0)
fig.colorbar(pcm5, ax=ax5, location="top", pad=0.01)
```

## Eigenfunction basis -- subsampling

```python
Phi2 = np.load(f"../data/ks_phi2_{NM}_{L}_{Lfact}_50.npy")
```

```python
phix = np.asarray([1, 2, 3, 4, 5, 6])

f0 = Phi2[:, phix[0]].reshape((N, M), order="C")
f1 = -Phi2[:, phix[1]].reshape((N, M), order="C")
f2 = Phi2[:, phix[2]].reshape((N, M), order="C")
f3 = Phi2[:, phix[3]].reshape((N, M), order="C")
f4 = -Phi2[:, phix[4]].reshape((N, M), order="C")
f5 = -Phi2[:, phix[5]].reshape((N, M), order="C")

fig, ((ax0, ax1, ax2), (ax3, ax4, ax5)) = plt.subplots(2, 3, layout="tight", figsize=[6.4, 6.4])

pcm0 = ax0.pcolormesh(xvec, nvec, f0, cmap=cmap_heat)
ax0.tick_params(axis="x", length=0)
ax0.tick_params(axis="y", length=0)
ax0.set_ylabel("Sample")
fig.colorbar(pcm0, ax=ax0, location="top", pad=0.01)

pcm1 = ax1.pcolormesh(xvec, nvec, f1, cmap=cmap_heat)
ax1.tick_params(axis="x", length=0)
ax1.tick_params(axis="y", length=0)
fig.colorbar(pcm1, ax=ax1, location="top", pad=0.01)

pcm2 = ax2.pcolormesh(xvec, nvec, f2, cmap=cmap_heat)
ax2.tick_params(axis="x", length=0)
ax2.tick_params(axis="y", length=0)
fig.colorbar(pcm1, ax=ax2, location="top", pad=0.01)

pcm3 = ax3.pcolormesh(xvec, nvec, f3, cmap=cmap_heat)
ax3.tick_params(axis="x", length=0)
ax3.tick_params(axis="y", length=0)
ax3.set_xlabel("Space")
ax3.set_ylabel("Sample")
fig.colorbar(pcm3, ax=ax3, location="top", pad=0.01)

pcm4 = ax4.pcolormesh(xvec, nvec, f4, cmap=cmap_heat)
ax4.tick_params(axis="x", length=0)
ax4.tick_params(axis="y", length=0)
fig.colorbar(pcm4, ax=ax4, location="top", pad=0.01)

pcm5 = ax5.pcolormesh(xvec, nvec, f5, cmap=cmap_heat)
ax5.tick_params(axis="x", length=0)
ax5.tick_params(axis="y", length=0)
fig.colorbar(pcm5, ax=ax5, location="top", pad=0.01)
```

## Sampled states

```python
uvec2 = np.load(f"../data/ks_true_{Ntrue}_{M}_{Lfact}_u.npy")[:N]
tvec2 = np.load(f"../data/ks_true_{Ntrue}_{M}_{Lfact}_t.npy")[:N]
tvec2 -= tvec2[0]
```

```python
samp1 = np.loadtxt(f"../data/ks_samples_{NM}_2048opt_50.dat", dtype=int)
t1 = np.empty((2048,), dtype=np.float64)
x1 = np.empty_like(t1)
for i in range(2048):
    t1[i] = (samp1[i] // M) * 4*dt
    x1[i] = (samp1[i] % M)/M * Ldom - Ldom/2
```

```python
fig, (ax0, ax1) = plt.subplots(1, 2, layout="tight", figsize=[4.8, 4.8], dpi=192)

pcm0 = ax0.pcolormesh(xvec, tvec2, uvec2, cmap=cmap_heat)
ax0.tick_params(axis="x", length=0)
ax0.tick_params(axis="y", length=0)
ax0.set_xlabel("Space")
ax0.set_ylabel("Time")
ax0.set_title("True", pad=35)
fig.colorbar(pcm0, ax=ax0, location="top", pad=0.01)

pcm1 = ax1.pcolormesh(xvec, tvec2, uvec2, cmap=cmap_heat)
ax1.plot(x1, t1, ".", c="k", markersize=0.7, alpha=0.7)
ax1.tick_params(axis="x", length=0)
ax1.tick_params(axis="y", length=0)
ax1.set_yticklabels([])
ax1.set_title(r"$r=2048$", pad=35)
fig.colorbar(pcm1, ax=ax1, location="top", pad=0.01)

#fig.savefig("samples1.png")
```

## State projection

```python
u = np.load(f"../data/ks_true_{Ntrue}_{M}_{Lfact}_u.npy")[:N]
t = np.load(f"../data/ks_true_{Ntrue}_{M}_{Lfact}_t.npy")[:N]
t -= t[0]

Phi1 = np.load(f"../data/ks_phi1_{NM}_{L}_{Lfact}_50.npy")
Phi2 = np.load(f"../data/ks_phi2_{NM}_{L}_{Lfact}_50.npy")

u = u.reshape(-1, order="C")
u1 = Phi1 @ (Phi1.T @ u)
u2 = Phi2 @ (Phi2.T @ u)

u = u.reshape((N, M), order="C")
u1 = u1.reshape((N, M), order="C")
u2 = u2.reshape((N, M), order="C")
```

```python
fig, (ax0, ax1, ax2) = plt.subplots(1, 3, layout="tight", figsize=[6.4, 4.8], dpi=192)

pcm0 = ax0.pcolormesh(xvec, t, u, cmap=cmap_heat)
ax0.tick_params(axis="x", length=0)
ax0.tick_params(axis="y", length=0)
ax0.set_xlabel("Space")
ax0.set_ylabel("Time")
ax0.set_title("True", pad=35)
fig.colorbar(pcm0, ax=ax0, location="top", pad=0.01)

pcm1 = ax1.pcolormesh(xvec, t, u1, cmap=cmap_heat)
ax1.tick_params(axis="x", length=0)
ax1.tick_params(axis="y", length=0)
ax1.set_yticklabels([])
ax1.set_title("Dilution", pad=35)
fig.colorbar(pcm1, ax=ax1, location="top", pad=0.01)

pcm2 = ax2.pcolormesh(xvec, t, u2, cmap=cmap_heat)
ax2.tick_params(axis="x", length=0)
ax2.tick_params(axis="y", length=0)
ax2.set_yticklabels([])
ax2.set_title("Subsampling", pad=35)
fig.colorbar(pcm1, ax=ax2, location="top", pad=0.01)

#fig.savefig("proj1a.png")
```

```python
fig, (ax0, ax1, ax2) = plt.subplots(1, 3, layout="tight", figsize=[6.4, 4.8], dpi=192)

pcm0 = ax0.pcolormesh(xvec, t, u, cmap=cmap_heat)
ax0.tick_params(axis="x", length=0)
ax0.tick_params(axis="y", length=0)
ax0.set_xlabel("Space")
ax0.set_ylabel("Time")
ax0.set_title("True", pad=35)
fig.colorbar(pcm0, ax=ax0, location="top", pad=0.01)

pcm1 = ax1.pcolormesh(xvec, t, u1-u, cmap=cmap_heat)
ax1.tick_params(axis="x", length=0)
ax1.tick_params(axis="y", length=0)
ax1.set_yticklabels([])
ax1.set_title("Dilution error", pad=35)
fig.colorbar(pcm1, ax=ax1, location="top", pad=0.01)

pcm2 = ax2.pcolormesh(xvec, t, u2-u, cmap=cmap_heat)
ax2.tick_params(axis="x", length=0)
ax2.tick_params(axis="y", length=0)
ax2.set_yticklabels([])
ax2.set_title("Subsampling error", pad=35)
fig.colorbar(pcm1, ax=ax2, location="top", pad=0.01)

print(np.sqrt(np.sum((u1-u)**2)) / np.sqrt(np.sum(u**2)) * 100)
print(np.sqrt(np.sum((u2-u)**2)) / np.sqrt(np.sum(u**2)) * 100)

#fig.savefig("proj1b.png")
```
